## Your final prompt

> Paste your complete, finished prompt below (double-click to edit).

```
please write a python program that tells a student wether they can afford to buy something based on their current income and savings. The program must include (inputs): the item name, Item price in AUD, Monthly income, Monthly fixed expenses (include rent, subscription, bills and debt) and finally their current savings total. Please also make sure to print the savings number (how much the student has left) after the purchase of the item.
```

````python
# Student Affordability Checker

# Get user inputs
item_name = input("Enter the item name: ")
item_price = float(input("Enter the item price in AUD: $"))

monthly_income = float(input("Enter your monthly income in AUD: $"))
monthly_expenses = float(input("Enter your monthly fixed expenses (rent, subscriptions, bills, debt) in AUD: $"))
current_savings = float(input("Enter your current savings total in AUD: $"))

# Calculate available money after monthly expenses
available_income = monthly_income - monthly_expenses

# Calculate total money available
total_available_money = available_income + current_savings

# Check if the student can afford the item
if total_available_money >= item_price:
    remaining_savings = current_savings - item_price

    print("\nPurchase Decision:")
    print(f"You can afford to buy the {item_name}.")
    print(f"Item price: ${item_price:.2f} AUD")
    print(f"Your remaining savings after purchase: ${remaining_savings:.2f} AUD")

else:
    amount_needed = item_price - total_available_money

    print("\nPurchase Decision:")
    print(f"You cannot afford to buy the {item_name} right now.")
    print(f"You need an additional ${amount_needed:.2f} AUD.")
    print(f"Your current savings remain: ${current_savings:.2f} AUD")

# Display financial summary
print("\nFinancial Summary:")
print(f"Monthly income: ${monthly_income:.2f} AUD")
print(f"Monthly fixed expenses: ${monthly_expenses:.2f} AUD")
print(f"Money available after expenses: ${available_income:.2f} AUD")
````
